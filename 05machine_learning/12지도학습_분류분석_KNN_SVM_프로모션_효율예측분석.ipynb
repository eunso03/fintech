```python
mem = pd.read_csv("./data/member.csv")
tran = pd.read_csv("./data/transaction.csv")
```

* 쇼핑몰 고객 데이터, 프로모션 쿠폰을 발행하고 사용 여부 데이터 수집

* F(Frequency) : 구매빈도
* M(Monetary) : 구매 금액
* 종속변수 : conversion => 고객이 프로모션에 반응 했는가? 1 = yes, 0 = no
