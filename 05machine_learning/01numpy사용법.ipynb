# 넘파이를 사용하는 이유
* 행렬 연산이 쉽고 빠르기 때문

```python
# 스칼라 연산
a = 10
b = 5
scalar = a + b
print(scalar)
```

```python
scalar + 10
```

```python
# 1차원 데이터 벡터의 연산
b1 = [1, 2, 3, 4 ,5]
# 리스트의 요소에 10을 곱한 후 다시 리스트로 만드세요.
result = []
for num in b1:
    result.append(num*10)
result
```

```python
result2 = [num * 10 for num in b1]
result2
```

```python
# 2차원 데이터 계산 *10 하기
second = np.array([[1, 2, 3], [1, 2, 3]])
print(second, second.ndim, second.shape)
```

```python
result = []
for num in second:
    mid_result = []
    for number in num:
        mid_result.append(number * 10)
    result.append(mid_result)
result = np.array(result)
```

```python
result.ndim
```

```python
result.shape
```

```python
result
```

```python
# 3차원 데이터의 요소에 10 곱하기
third = np.array([[[1, 2, 3], [1, 2, 3]], [[1, 2, 3], [1, 2, 3]]])
print(third, third.ndim, third.shape)
```

```python
result = []
for num1 in third:
#     print(num1, end="\n\n")
    num1_result = []
    for num2 in num1:
        num2_result = []
        for num3 in num2:
#             print(num2)
            num2_result.append(num3 * 10)
        num1_result.append(num2_result)
    result.append(num1_result)
result = np.array(result)
```

```python
result.ndim
```

```python
result.shape
```

```python
result
```

```python
# 4차원 데이터에 10 곱하기
forth = np.array([[[[1, 2, 3], [1, 2, 3]], [[1, 2, 3], [1, 2, 3]]],
                 [[[1, 2, 3], [1, 2, 3]], [[1, 2, 3], [1, 2, 3]]]])
print(forth, forth.ndim, forth.shape)
```
