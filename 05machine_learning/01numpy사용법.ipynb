# 넘파이를 사용하는 이유
* 행렬 연산이 쉽고 빠르기 때문

```python
# 스칼라 연산
a = 10
b = 5
scalar = a + b
print(scalar)
```

```python
scalar + 10
```

```python
# 1차원 데이터 벡터의 연산
b1 = [1, 2, 3, 4 ,5]
# 리스트의 요소에 10을 곱한 후 다시 리스트로 만드세요.
result = []
for num in b1:
    result.append(num*10)
result
```

```python
result2 = [num * 10 for num in b1]
result2
```

```python
# 2차원 데이터 계산 *10 하기
second = np.array([[1, 2, 3], [1, 2, 3]])
print(second, second.ndim, second.shape)
```

```python
result = []
for num in second:
    mid_result = []
    for number in num:
        mid_result.append(number * 10)
    result.append(mid_result)
result = np.array(result)
```

```python
result.ndim
```

```python
result.shape
```

```python
result
```

```python
# 3차원 데이터의 요소에 10 곱하기
third = np.array([[[1, 2, 3], [1, 2, 3]], [[1, 2, 3], [1, 2, 3]]])
print(third, third.ndim, third.shape)
```

```python
result = []
for num1 in third:
#     print(num1, end="\n\n")
    num1_result = []
    for num2 in num1:
        num2_result = []
        for num3 in num2:
#             print(num2)
            num2_result.append(num3 * 10)
        num1_result.append(num2_result)
    result.append(num1_result)
result = np.array(result)
```

```python
result.ndim
```

```python
result.shape
```

```python
result
```

```python
# 4차원 데이터에 10 곱하기
forth = np.array([[[[1, 2, 3], [1, 2, 3]], [[1, 2, 3], [1, 2, 3]]],
                 [[[1, 2, 3], [1, 2, 3]], [[1, 2, 3], [1, 2, 3]]]])
print(forth, forth.ndim, forth.shape)
```

* np.where(조건, 참, 거짓) 3항 연산

```python
num = np.arange(1, 11)
num
```

```python
np.where(num % 2 == 0, '짝수', '홀수')
```

```python
result = []
for i in num:
    if i % 2 == 0:
        result.append('짝수')
    else:
        result.append('홀수')
result
```

# 배열 합치기 np.concatenate()

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(a, a.shape, a.ndim)
print(b, b.shape, b.ndim)
```

* np.concatenate((a, b))
* 합치는 기본 방향은 axis=0 행이 늘어남, 아래로 붙음
* axis=1 열이 늘어남, 옆으로 붙음

```python
np.concatenate((a, b))
```

```python
np.concatenate((a, b), axis=1)
```

# 합치는 방향의 모양이 일치하지 않으면 합치기 X

```python
c = np.array([5, 6])
print(c, c.ndim, c.shape)
```

```python
np.concatenate((a, c))
```

```python
d = np.array([[7, 8, 9]])
print(d, d.ndim, d.shape)
```

```python
np.concatenate((a, d), axis=1)
```

```python
e = np.array([[1, 2, 3], [4, 5, 6]])
print(e, e.ndim, e.shape)
```

```python
np.concatenate((d, e))
```

```python
f = np.array([[1, 2], [3, 4], [5, 6]])
print(f, f.ndim, f.shape)
```

```python
np.concatenate((e, f))
```

```python
# 전치행렬: 행과 열을 바꿈
print(f)
print(f.shape)
print(f.T)
print(f.T.shape)
```

```python
np.concatenate((e, f.T), axis=1)
```
