# print문의 옵션
* sep: 구분자 - print문 안에 여러 개의 자료를 구분해주는 문자
* 기본 값은 " " 공백

```python
print("이것은", '파이썬', '입니다.')
```

```python
print("이것은", '파이썬', '입니다.', sep="_")
```

```python
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
```

* end 옵션 - print가 끝나고 실행될 옵션
* 기본 값은 \n - 이스케이프 문자 new line(새 줄)

```python
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
```

```python
# end값을 공백 " " 으로 준 경우
print("헐크", '아이언맨', '스파이더맨', sep=" vs ", end=" ")
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
```

```python
# end값을 공백 "\t"(tap) 으로 준 경우
print("헐크", '아이언맨', '스파이더맨', sep=" vs ", end="\t")
print("헐크", '아이언맨', '스파이더맨', sep=" vs ")
```

```python
# end값을 공백 "\r"(carrige return)으로 준 경우
print("헐크", '아이언맨', '스파이더맨', sep=" vs ", end="\r")
print("스파이더맨", '아이언맨', sep=" vs ")
```

```python
# end값을 공백 "\n"(new line)으로 준 경우
print("헐크", '아이언맨', '스파이더맨', sep=" vs ", end="\n\n\n")
print("스파이더맨", '아이언맨', '헐크', sep=" vs ")
```

```python
# end값을 공백 "\n"(new line)으로 준 경우
print("헐크", 1, 3.14, sep="-", end="\n\n\n")
print("스파이더맨", '아이언맨', '헐크', sep=" vs ")
```
