* 점수를 입력받아서 82-100 A, 61-80 B, 41-60 C, 21-40 D, 0-20 F등급

```python
score = int(input("점수를 입력하세요"))
if 0 <= score <=100:
    if 0 <= score <= 20:
        print(f"{score}는 F등급 입니다.")
    elif 21 <= score <=40:
        print(f"{score}는 D등급 입니다.")
    elif 41 <= score <=60:
        print(f"{score}는 C등급 입니다.")
    elif 61 <= score <=80:
        print(f"{score}는 B등급 입니다.")
    else:
        print(f"{score}는 A등급 입니다.")
else:
    print("숫자를 0-100사이로 입력해주세요")
```

* 1-50사이의 숫자를 입력방아서 3의 배수면 3의 배수입니다, 3의 배수가 아니면 3의 배수가 아닙니다.
    * 입력받기, 정수로 변환하기
    * 1-50 사이의 숫자인가?
    * 3의 배수인가? % 3 ==0
    * 출력

```python
num = int(input("정수로 숫자를 입력하세요"))
# print(num, type(num))
if 1 <= num <= 50:
    # print("숫자가 1-50사이의 정수입니다.")
    if num % 3 == 0:
        print(f"{num}은 3의 배수입니다")
    else:
        print(f"{num}은 3의 배수가 아닙니다.")
else: 
    print("숫자를 1-50사이의 정수로 입력해주세요")
```

## 3. 조건 표현식
* 참일 때 실행 코드 if 조건 else 거짓일 때 실행 코드

```python
# 일반적인 if else문
num = int(input("숫자를 입력하세요"))
# print(num, type(num))
if num % 2 == 0:
    print("짝수")
else:
    print("홀수")
```

```python
# 조건 표현식
num = int(input("숫자를 입력하세요"))
# print(num, type(num))
print("짝수") if num % 2 == 0 else print("홀수")
```
