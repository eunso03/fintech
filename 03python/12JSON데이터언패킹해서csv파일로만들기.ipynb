```python
import json
import pandas as pd
```

```python
with open("./data/com_kbstar_kbbank.json", "r", encoding = "utf-8") as file:
    data = file.read()
```

* 방법1

```python
user = []
star = []
review = []
date = []
respond = []
```

```python
for i in reviews:
    user.append(i["사용자"])
    star.append(i["별점"])
    review.append(i["리뷰"])
    date.append(i["날짜"])
    respond.append(i["개발자 응답"])
    
review_info = dict(사용자=user, 별점=star, 리뷰=review, 날짜=date, 개발자응답=respond)
review_info
```

```python
import pandas as pd
```

```python
review_df = pd.DataFrame(review_info)
review_df
```

* 방법2

```python
result = {}
for key in reviews[0]:
    temp = []
    for review in reviews:
        temp.append(review[key])
    result[key] = temp
    
len(result['별점'])
```

```python
df = pd.DataFrame(result)
df
```

별점은 숫자로 변환해서 넣기, 나머지는 특수문자 제거하기

```python
reviews[0]["리뷰"].replace("?", " ")
```

```python
result = {}
for key in reviews[0]:
    temp = []
    for review in reviews:
        if key == "별점":
            temp.append(int(review[key]))
        else:
            temp2 = str(review[key]).replace("!", " ").replace(".", " ").replace(",", " ").replace("(", " ").replace(")", " ")
            temp.append(temp2)
    result[key] = temp
    
len(result['별점'])
```

```python
df = pd.DataFrame(result)
df
```

```python
df.info()
```

* 방법3

```python
result2 = {}
for review in reviews:
    for key, value in review.items():
        result2.setdefault(key, []).append(value)
```

```python
result2_df = pd.DataFrame(result2)
result2_df
```

전국치킨집리스트

```python
with open("./data/전국치킨집리스트.json", "r", encoding = "utf-8") as file2:
    data2 = file2.read()
```

```python
data = json.loads(data2)
len(data)
```

* data에는 리스트 요소가 37개 있음
* 리스트 요소 1개에는 1000개의 가게 정보가 있음
* 가게 정보는 body[0]["body"]["items"]에 있다. 
* 가게 정보를 구성하는 key는 39개이다.

```python
data[0]["body"]["items"][0]
```

```python
len(data[0]["body"]["items"][0])
```

```python
result3 = {}
for page in data:
    for item in page["body"]["items"]:
        for key, value in item.items():
            result3.setdefault(key, []).append(value)
    #     print(key, value)
len(result3["bizesId"])
```

```python
df = pd.DataFrame(result3)
df
```

```python
len(data[0]["header"]["columns"])
```

```python
result4 = {}
for page in data:
    for item in page["body"]["items"]:
        for key, value in zip(page["header"]["columns"], item.values()):
            result4.setdefault(key, []).append(value)
```

```python
df = pd.DataFrame(result4)
df
```
