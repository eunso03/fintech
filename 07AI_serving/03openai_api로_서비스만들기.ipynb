# gradio를 이용한 openAI 챗봇

```python
import os
import gradio as gr
from openai import OpenAI
# from dotenv import load_dotenv

# 🔐 환경변수 로드
# load_dotenv()
# api_key = os.getenv("OPENAI_API_KEY")

# 🧠 OpenAI 클라이언트 초기화
client = OpenAI()

# ✨ 응답 생성 함수
def ask_gpt(input_text):
    try:
        response = client.responses.create(
            model="gpt-4o-mini",
            input=input_text,
            instructions="친절한 파이썬 강사처럼 설명해 주세요.",
        )
        return response.output_text
    except Exception as e:
        return f"⚠️ 오류 발생: {str(e)}"

# 🎨 Gradio UI 구성
with gr.Blocks() as demo:
    gr.Markdown("## 🤖 GPT-4o-mini 기반 파이썬 강사 챗봇")

    with gr.Row():
        input_text = gr.Textbox(label="질문을 입력하세요", placeholder="예: Python에서 리스트란?")
        output_text = gr.Textbox(label="GPT의 답변", lines=8, interactive=False)

    input_text.submit(ask_gpt, inputs=input_text, outputs=output_text)

demo.launch()
```
